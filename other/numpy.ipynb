```python
#Sol
output = np.ones((5,5),dtype = 'int8')
print(output)

z  =np.zeros((3,3),dtype = 'int8')
print(z)
z[1,1] = 9
print(z)
output[1:-1, 1:-1] = z
print(output)
```

```python
#be careful when copying arrays
a = np.array([1,2,3])
b = a
print(b)
b[0] = 100
print(b)
print(a)
```

```python
#resolve
a = np.array([1,2,3])
b = a.copy()
print(b)
b[0] = 100
print(b)
print(a)
```

### reorganizing arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)
```

```python
before.shape
```

```python
#reshape
after = before.reshape(8,1)
print(after,after.shape)
after = before.reshape(2,2,2)
print(after)
```

```python
#vertical stacking arrays
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

np.vstack([v1,v2,v2,v2])
```

```python
#horizontal stack

h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack([h1,h2])
```
