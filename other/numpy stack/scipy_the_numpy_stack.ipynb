```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from scipy.stats import norm
```

```python
x = np.linspace(-6,6,100)
```

# pdf cdf and logpdf

```python
fx = norm.pdf(x,loc = 0,scale = 1)
```

```python
plt.plot(x,fx);
```

```python
ffx = norm.cdf(x,loc=0,scale=1)
```

```python
 plt.plot(x,ffx)
```

```python
logfx = norm.logpdf(x,loc=0,scale = 1)
plt.plot(x,logfx);
```

# convolution

```python
from PIL import Image
```

```python
im = Image.open('/content/sample_data/WhatsApp Image 2022-02-23 at 8.33.58 PM.jpeg')
```

```python
gray = np.mean(im, axis = 2)
```

```python
x = np.linspace(-6,6,50)
fx = norm.pdf(x,loc = 0,scale =1)
```

# GAussian filter

image.png

```python
filt = np.outer(fx,fx)
```

```python
plt.imshow(filt)
```

```python
plt.imshow(filt, cmap= 'gray')
```

```python
from scipy.signal import convolve2d
out = convolve2d(gray,filt)
```

```python
plt.figure(figsize=(16,8))
plt.subplot(1,2,1)#r,c,p
plt.imshow(gray,cmap='gray')
plt.subplot(1,2,2)#r,c,p
plt.imshow(out,cmap='gray');
```
