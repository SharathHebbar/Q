## Speed test

```python
from datetime import datetime

a = np.random.randn(100)
b = np.random.randn(100)
t= 100000
```

```python
def slow_dot_product(a,b):
  r = 0
  for i,j in zip(a,b):
    r+=i*j
  return r

t0 = datetime.now()
for i in range(t):
  slow_dot_product(a,b)
d0 = datetime.now()-t0

t0 = datetime.now()
for i in range(t):
  a.dot(b)
d1 = datetime.now()-t0

print(d0.total_seconds()/d1.total_seconds())
```

## Speed test for matrix multiplication

```python
import numpy as np
from datetime import datetime

a = np.random.randn(100,100)
b = np.random.randn(100,100)
c = np.zeros((100,100))

a.shape,b.shape,c.shape
```

```python
def slow_matrix_multi(a,b,c):
  for i in range(len(a)):
    for j in range(len(a[0])):
      for k in range(len(b)):
        c[i][j] += a[i][k] * b[k][j]
  return c

t0 = datetime.now()

c1 = slow_matrix_multi(a,b,c)
d0 = datetime.now()-t0

t0 = datetime.now()
c2 = a.dot(b)
d1 = datetime.now()-t0
print("c1: ",c1)
print("c2: ",c2)
print(d0.total_seconds()/d1.total_seconds())
```
